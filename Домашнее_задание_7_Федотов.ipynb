Задание 1

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)
response.raise_for_status()

posts = response.json()

for post in posts[:5]:
    print("Заголовок:", post["title"])
    print("Текст:", post["body"])
    print(" ")
```

```python
import requests

API_KEY = "ddff592ad2b05fc8f0b1ba0721a26178"
url = "https://api.openweathermap.org/data/2.5/weather"

city = input("Введите название города: ")

params = {
    "q": city,
    "appid": API_KEY,
    "units": "metric",
    "lang": "ru"
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()
    temp = data["main"]["temp"]
    description = data["weather"][0]["description"]
    print(f"Температура в городе {city}: {temp}°C")
    print(f"Погода: {description}")
elif response.status_code == 404:
    print("Город не найден")
elif response.status_code == 401:
    print("Неверный или ещё не активированный API-ключ")
else:
    print("Ошибка:", response.status_code)
```
